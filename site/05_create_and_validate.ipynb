# Exercise 5 — Create and validate (optional)

**Goal:** create an IFC object and check whether it contains a required piece of information. Allow 20 minutes. This is a small independent model, so it does not change the supplied sample.

An information check asks whether a required value is present and has the expected form. It does **not** verify the physical or structural performance of the object.

```python
from compas_ifc.bim import BuildingInformationModel

model = BuildingInformationModel.template(schema="IFC4", unit="m")
storey = model.storeys[0]
print("New model:", model.name)
print("Storey:", storey.name)
```

## 1. State an information requirement

Our simple rule says each wall must have a `CourseInfo` property set with a text `note`. The `Specification` describes the rule; `validate` reports whether objects satisfy it.

```python
from pydantic import BaseModel
from compas_ifc.validation import Specification

class CourseInfo(BaseModel):
    note: str

rule = Specification(
    name="Course note",
    ifc_types=["IfcWall"],
    required_psets={"CourseInfo": CourseInfo},
)
```

## 2. Create a wall with its information

`create_wall` places the new object under the storey. For this information exercise, it does not need geometry. Try changing the note to your own short explanation.

```python
wall = model.create_wall(
    name="Practice wall",
    parent=storey,
    properties={"CourseInfo": {"note": "Created in Binder for learning"}},  # CHANGE the note
)
print(wall.ifc_type, wall.name, wall.global_id)
print(wall.properties)
```

## 3. Check the result

The result says `pass` or `fail` for this specified information rule. It is a narrow check, not a general IFC or engineering approval.

```python
for result in model.validate([rule]):
    print(result.element_name, result.status, result.missing_psets, result.property_errors)
```
